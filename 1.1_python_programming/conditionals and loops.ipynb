```python
print("school grading system")
score=int(input('score: '))
if score > 90:
    print("A+ grade")
elif score >= 80 and score <= 90:
    print("A grade")
elif score >= 70 and score <=70:
    print("B+ grade")
elif score >= 60 and score <=69:
    print("B grade")
elif score >= 50 and score <=59:
    print("c grade")
else:
    print("failed")
```
