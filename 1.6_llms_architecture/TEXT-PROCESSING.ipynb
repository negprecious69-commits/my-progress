<h3>TEXT CLEANING USING REGULAR EXPRESSION</h3>

```python
import re
white_space_splitting = re.split(r'\s', raw_text)
print(white_space_splitting)
```

```python
#remove @mentions and #tags
raw_text = re.sub(r'@\w+'," ",raw_text,flags=re.MULTILINE)
raw_text = re.sub(r'#\w+'," ",raw_text,flags=re.MULTILINE)

# remove anything that is not alphabet
raw_text = re.sub(r'[^a-zA-Z\s]','',raw_text)

#remove anything whitespace
raw_text = re.sub(r'\s+',' ',raw_text).strip()

raw_text
```

```python
# splitting using other criteria
white_space_splitting_with_special_xters = re.split(r'--|[@3^!#*?&]|\s', raw_text)
print(white_space_splitting_with_special_xters)
```

<h3>remove white spaces from the splited text</h3>

```python
strip_text = [word.strip() for word in white_space_splitting_with_special_xters if word.strip()]
len(strip_text)
```
