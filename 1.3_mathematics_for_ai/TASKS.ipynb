```python
query = np.array([3, 4])
key   = np.array([4, 3]) 
b=np.array([])
total=sum(x*y for x,y in zip(query,key))
print(total)
```

```python
fig , ax=plt.subplots(figsize=(6,4))

ax.quiver(0, 0, query[0], query[1], angles='xy', scale_units='xy', scale=1, color='navy', label='query = [3, 4]')
ax.quiver(0, 0, key[0], key[1], angles='xy', scale_units='xy', scale=1, color='gold', label='key = [4, 3]')
plt.gca().set_aspect('equal')
plt.grid(True)
plt.legend()
ax.set_xlim(-1,6)
ax.set_ylim(-1,6)
```

```python
#for the absolute value
vq=np.linalg.norm(query)
vk=np.linalg.norm(key)
v=vq*vk
print(v)
```

```python
costheta=total/v
print(costheta)
```

```python
theta=np.arccos(costheta)
theta
```

```python
angle=np.degrees(theta)
angle
```

```python
fig , ax=plt.subplots(figsize=(6,4))

ax.quiver(0, 0, query[0], query[1], angles='xy', scale_units='xy', scale=1, color='navy', label='query = [3, 4]')
ax.quiver(0, 0, key[0], key[1], angles='xy', scale_units='xy', scale=1, color='gold', label='key = [4, 3]')
plt.gca().set_aspect('equal')
plt.grid(True)
plt.legend()
ax.set_xlim(-1,6)
ax.set_ylim(-1,6)
plt.text(2, 1, f'angle = {angle:.1f} degrees')
plt.show()
```

<h3>5</h3>

```python
v=np.array([3, 4])

# computing the norm
norm=np.linalg.norm(v)
print("Norm:", norm)

# computing the unit vector
unit_vector=v/norm
print("unit vector:", unit_vector)

# checking the norm
print("norm of the unit vector:", np.linalg.norm(unit_vector))
```

```python
def normalize(vec):
    norm=np.linalg.norm(vec)
    return vec/norm
print(normalize(v))    
```

```python
import matplotlib.pyplot as plt

unit_vector=normalize(v)

fig , ax=plt.subplots(figsize=(6,4))

ax.quiver(0, 0, v[0], v[1], angles='xy', scale_units='xy', scale=1, color='navy', label='v = [3, 4]')
ax.quiver(0, 0, unit_vector[0], unit_vector[1], angles='xy', scale_units='xy', scale=1, color='gold', label='unit_vector = [6, 8')
plt.gca().set_aspect('equal')
plt.grid(True)
plt.legend()
ax.set_xlim(-1,6)
ax.set_ylim(-1,6)
plt.text(2, 1, f'angle = {angle:.1f} degrees')
plt.show()
```

<h3>6</h3>

```python
v = np.array([3, -4, 12])

l2_norm=np.linalg.norm(v)
print("l2 norm:", l2_norm)

manual_l2=np.sqrt(np.sum(v**2))
print("Manual l2 norm:", manual_l2)

v**2

np.sum(v**2)

np.sqrt(25)

print(l2_norm == manual_l2)
```

```python
l1_norm=np.linalg.norm(v, ord=1)
manual_l1=np.sum(np.abs(v))

print("l1 using numpy:", l1_norm)
print("l1 manually:", manual_l1)
```

```python
# Generating 2000 random 2D points
points=np.random.uniform(-2, 2, (2000, 2))

#compute l2 norm
l2=np.linalg.norm(points, axis=1)

# Compute l1 norm
l1=np.linalg.norm(points, ord=1, axis=1)

#First plot (l2)
plt.subplot(1, 2, 1)
plt.scatter(points[:,0], points[:,1], c=l2)
plt.title("l2 Norm")

#Second plot (l1)
plt.subplot(1, 2, 2)
plt.scatter(points[:,0], points[:,1], c=l1)
plt.title("l1 Norm")

plt.show()
```

```python
Q=np.array([[1, 2], [3, 4]])

K=np.array([[5, 6], [7, 8]])

result=Q+K
print(result)

result=Q*K
print(result)

print(K.T)

result=Q@K.T
print(result)
```

```python
print(np.matmul(Q, K.T))
```

```python
# DETERMINANT
det=np.linalg.det(Q)
print(det)

# inverse 
inverse=np.linalg.inv(Q)
print(inverse)

print(Q@inverse)
```
