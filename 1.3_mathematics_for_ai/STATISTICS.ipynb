```python
# HOW TO CREATE A MATRIX IN NUMPY

import numpy as np

A = np.array([
    [1, 2, 3, 4],
    [5, 6, 7, 8],
    [6, 10, 11, 12]
])

print(A.shape) # 3 rows and 4 col
print(A.ndim) # dimension
print(A.size) # total elements
print(A.dtype)
```

```python
# slicing element: A[row, col]
print(A[0, 0]) # top left corner of first number
print(A[1, 2]) # row 1 col 2
print(A[-1, -1]) # bottom left corner


# SLICING AN entire row
print(A[1, :])
print(A[1])
```

```python


import numpy as np
import matplotlib.pyplot as plt
m=np.random.rand(5, 5)
plt.figure(figsize=(6,5))
plt.imshow(m, cmar = np.random.rand(3, 3) )# generates numbers btw 0 and 1
print(rp='Blues')
plt.colorbar()
plt.title('Matrix as a Heatmap')

for i in range(5):
    for j in range(5):
        plt.text(j, i, f"{m[i,j]:.2f}",
            ha='center', va='center', fontsize=9)
```
