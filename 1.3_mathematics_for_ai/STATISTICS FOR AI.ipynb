<h3>mean<h3\>

```python
import numpy as np
# calculating the mean
prices = np.array([210, 225, 240, 900])
print("Mean:", np.mean(prices))
print("Mean (no outlier):", np.mean(prices[:-1]))
```

<h3>median<h3\>

```python
# deals with outliers
prices = np.array([210, 225, 240, 900])
print("Median:", np.median(prices))
even_data = np.array([1, 2, 3, 4])
print("Median (no outlier):", np.median(even_data))
```

<h3>mode</h3>

```python
# deals with categorical data, missing values like color
shoe_sizes = np.array([8, 9, 9, 10, 9, 11, 8, 9])
np.unique(shoe_sizes)
values, counts=np.unique(shoe_sizes, return_counts=True)
mode_value=values[np.argmax(counts)]
print("Mode:", mode_value, '! appears', counts.max(), "times")
```

<h3>Variance and standard deviation</h3>

```python
scores=np.array([60, 70, 80, 90, 100])
mean=np.mean(scores)
deviations=scores-mean           # vectorized: substracts mean from every element
sq_deviation=deviations**2       # squares each deviation (removes negative signs)
variance=np.var(scores, ddof=1)
std_dev=np.std(scores, ddof=1)   # ddof -> divide by (n-1), sample variance
print("variance:", variance)
print("std Dev:", std_dev.round(2))
```
