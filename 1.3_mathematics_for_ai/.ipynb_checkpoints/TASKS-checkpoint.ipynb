<h3>4</h3>

```python
query = np.array([3, 4])
key   = np.array([4, 3]) 
b=np.array([])
total=sum(x*y for x,y in zip(query,key))
print(total)
```

```python
fig , ax=plt.subplots(figsize=(6,4))

ax.quiver(0, 0, query[0], query[1], angles='xy', scale_units='xy', scale=1, color='navy', label='query = [3, 4]')
ax.quiver(0, 0, key[0], key[1], angles='xy', scale_units='xy', scale=1, color='gold', label='key = [4, 3]')
plt.gca().set_aspect('equal')
plt.grid(True)
plt.legend()
ax.set_xlim(-1,6)
ax.set_ylim(-1,6)
```

```python
#for the absolute value
vq=np.linalg.norm(query)
vk=np.linalg.norm(key)
v=vq*vk
print(v)
```

```python
costheta=total/v
print(costheta)
```

```python
theta=np.arccos(costheta)
theta
```

```python
angle=np.degrees(theta)
angle
```

```python
fig , ax=plt.subplots(figsize=(6,4))

ax.quiver(0, 0, query[0], query[1], angles='xy', scale_units='xy', scale=1, color='navy', label='query = [3, 4]')
ax.quiver(0, 0, key[0], key[1], angles='xy', scale_units='xy', scale=1, color='gold', label='key = [4, 3]')
plt.gca().set_aspect('equal')
plt.grid(True)
plt.legend()
ax.set_xlim(-1,6)
ax.set_ylim(-1,6)
```
